## Code 04 — 조건문

**예상 출력**

```text

```

## Code 05 — 반복문과 누적

**예상 출력**

```text

```


````python
numbers = [2, 4, 6, 8]
total = 0

for number in numbers:
    total += number

print(total)
````

## Code 06 — 함수

## Code 08 — 리스트 컴프리헨션

**예상 출력**

```text

```


````python
numbers = [1, 2, 3, 4, 5]
squares = [number ** 2 for number in numbers]
print(squares)
````

## Code 09 — 문자열과 반복문

**예상 출력**

```text
[2, 5, 6]
```


````python
words = ["AI", "helps", "coding"]
lengths = []

for word in words:
    lengths.append(len(word))

print(lengths)
````

## Code 10 — 함수와 조건문

**예상 출력**

```text
['odd', 'even', 'odd', 'even']
```


````python
def classify_number(number):
    if number % 2 == 0:
        return "even"
    return "odd"

results = [classify_number(number) for number in [3, 8, 11, 14]]
print(results)
````
